## ۴. اجرای سیستم تحلیل (بدون UI)
در این بخش می‌توانید به صورت تعاملی نظرات خود را تحلیل کنید.

```python
from processor import SentimentRAG

print("در حال بارگذاری مدل‌ها (ممکن است چند دقیقه طول بکشد)...")
rag = SentimentRAG()

def start_chat():
    print("\n--- سامانه تحلیل هوشمند نظرات دیجی‌کالا ---")
    print("برای خروج عبارت 'exit' را وارد کنید.\n")
    
    while True:
        user_input = input("نظر خود را وارد کنید: ")
        if user_input.lower() == 'exit':
            break
            
        if not user_input.strip():
            continue
            
        print("\nدر حال تحلیل...")
        score, confidence = rag.get_sentiment(user_input)
        explanation = rag.generate_explanation(user_input, score)
        
        label = "مثبت" if score > 3 else "منفی" if score < 3 else "خنثی"
        
        print(f"\nوضعیت: {label} (امتیاز: {score} از ۵)")
        print(f"ضریب اطمینan: {confidence:.2%}")
        print(f"دلیل و استدلال: {explanation}")
        print("-" * 50 + "\n")

start_chat()
```
