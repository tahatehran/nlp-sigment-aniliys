## ۴. تست سیستم RAG (بدون رابط کاربری)
می‌توانید به صورت مستقیم کلاس پردازشگر را تست کنید.

```python
from processor import SentimentRAG

print("در حال بارگذاری مدل‌ها (ممکن است چند دقیقه طول بکشد)...")
rag = SentimentRAG()

test_text = "این گوشی واقعا عالیه، سرعتش خیلی بالاست اما قیمتش کمی تنده."
score, confidence = rag.get_sentiment(test_text)
explanation = rag.generate_explanation(test_text, score)

print("--- نتیجه تحلیل ---")
print(f"متن: {test_text}")
print(f"امتیاز: {score}/5 (اطمینان: {confidence:.2%})")
print(f"توضیح هوش مصنوعی: {explanation}")
```
